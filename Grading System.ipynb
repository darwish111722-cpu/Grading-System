```python


from abc import ABC, abstractmethod



class person(ABC):
    def __init__(self, id, name, email):
        self.__id = id
        self.__name = name
        self.__email = email

    def get_id(self):
        return self.__id

    def set_id(self, id):
        self.__id = id

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_email(self):
        return self.__email

    def set_email(self, email):
        self.__email = email

    @abstractmethod
    def get_info(self):
        pass
```

```python
class student(person):
    def __init__(self, id, name, email):
        super().__init__(  id, name, email)
        self.__grade = []
        self.__Arabic=0
        self.__Math=0
        self.__English=0
        self.__Chemistry=0
        self.__Physics=0




    def add_grade(self, Arabic , English   ,  Math,   Chemistry,   Physics):


        self.__Arabic = Arabic
        self.__Math = Math
        self.__English = English
        self.__Chemistry = Chemistry
        self.__Physics = Physics

        self.__grade.append(Arabic)
        self.__grade.append(English)
        self.__grade.append(Math)
        self.__grade.append(Chemistry)
        self.__grade.append(Physics)

    def calculate_grade(self):
        return (sum(self.__grade) / 500)*100
    def get_result(self):

        if self.calculate_grade()>=90:
            x="A"
        elif self.calculate_grade()>=75:
            x="B"
        elif self.calculate_grade()>=60:
            x="C"
        elif self.calculate_grade()>=50:
            x="D"
        else:
            x="F"


        return f"Arabic:{self.__Arabic}\nEnglish:{self.__English}\nMath:{self.__Math}\nChemistry:{self.__Chemistry}\nPhysics:{self.__Physics}\npercentage:{self.calculate_grade()}%\nGrade:{x}"


    def get_info(self):
        return f"id: {self.get_id() }\nName: {self.get_name()}\nEmail: {self.get_email()}\nGrades: {self.get_result()}"
```

```python
class Teacher(person):
    def __init__(self, id, name, email,subject):
        super().__init__(id, name, email)
        self.__subject = subject
    def get_subject(self):
        return self.__subject
    def set_subject(self, subject):
        self.__subject = subject
    def get_info(self):
        return f"id: {self.get_id()}\nName: {self.get_name()}\nEmail: {self.get_email()}\nSubject: {self.get_subject()}"
```

```python
class GradingSystem :
    def __init__(self):
        self.__students = []
        self.__teachers = []
    def add_student(self, student):
        self.__students.append(student)
    def add_teacher(self, teacher):
        self.__teachers.append(teacher)
    def Search_Student_By_ID(self, id):
        for x in self.__students:
            if x.get_id()==id:
                return x.get_name()
        return "Not Found"

    def Search_Teacher_By_ID(self, id):
        for x in self.__teachers:
            if x.get_id()==id:
                return x.get_name()
        return "Not Found"

    def Delete_Student_By_ID(self, id):
        for x in self.__students:
            if x.get_id()==id:
                self.__students.remove(x)
                return "Student Deleted"
        return "Not Found"
    def Delete_Teacher_By_ID(self, id):
        for x in self.__teachers:
            if x.get_id()==id:
                self.__teachers.remove(x)
                return "Teacher Deleted"
        return "Not Found"

    def get_all_students(self):
        for x in self.__students:
            print('*'*30)
            print(x.get_info())
    def get_all_teachers(self):
        for x in self.__teachers:
            print('*'*30)
            print(x.get_info())
    def update_Name_Student_By_ID(self, id, name):
        for x in self.__students:
            if x.get_id() == id:
                x.set_name(name)
                return "Student Updated"
        return "Not Found"
    def update_Name_Teacher_By_ID(self, id, name):
        for x in self.__teachers:
            if x.get_id() == id:
                x.set_name(name)
                return "Teacher Updated"
        return "Not Found"

    def update_Email_Student_By_ID(self, id, email):
        for x in self.__students:
            if x.get_id() == id:
                x.set_email(email)
                return "Student Updated"
        return "Not Found"

    def update_Email_Teacher_By_ID(self, id, email):
        for x in self.__teachers:
            if x.get_id() == id:
                x.set_email(email)
                return "Teacher Updated"
        return "Not Found"
    def update_ID_Student_By_Name(self, name, id):
        for x in self.__students:
            if x.get_name().lower() == name.lower():
                x.set_id(id)
                return "Student Updated"
        return "Not Found"
    def update_ID_Teacher_By_Name(self, name, id):
        for x in self.__teachers:
            if x.get_name().lower() == name.lower():
                x.set_id(id)
                return "Teacher Updated"
        return "Not Found"

    def update_Subject_Teacher_By_ID(self, id, subject):
        for x in self.__teachers:
            if x.get_id() == id:
                x.set_subject(subject)
                return "Teacher Updated"
        return "Not Found"
```

```python

import tkinter as tk
from tkinter import messagebox

system = GradingSystem()


BG = "#0F172A"
SIDE = "#111827"
CARD = "#1E293B"
BLUE = "#3B82F6"
GREEN = "#10B981"
RED = "#EF4444"
TEXT = "#F8FAFC"
MUTED = "#94A3B8"
```

```python



# ---------- Helpers ----------
def clear():
    for w in content.winfo_children():
        w.destroy()


def title(txt, sub=""):
    tk.Label(content, text=txt, bg=BG, fg=TEXT,
             font=("Segoe UI", 24, "bold")).pack(anchor="w", pady=(10, 2))
    if sub:
        tk.Label(content, text=sub, bg=BG, fg=MUTED,
                 font=("Segoe UI", 10)).pack(anchor="w", pady=(0, 20))


def field(parent, text):
    tk.Label(parent, text=text, bg=CARD, fg=MUTED,
             font=("Segoe UI", 10)).pack(anchor="w", padx=20, pady=(12, 3))
    e = tk.Entry(parent, bg="#334155", fg=TEXT, insertbackground=TEXT,
                 relief="flat", font=("Segoe UI", 11))
    e.pack(fill="x", padx=20, ipady=8)
    return e


def button(parent, text, command, color=BLUE):
    tk.Button(parent, text=text, command=command, bg=color, fg="white",
              activebackground=color, activeforeground="white",
              relief="flat", bd=0, font=("Segoe UI", 10, "bold"),
              cursor="hand2", padx=15, pady=9).pack(fill="x", padx=20, pady=7)


def output(text):
    result.config(state="normal")
    result.delete("1.0", "end")
    result.insert("end", text)
    result.config(state="disabled")


def card():
    f = tk.Frame(content, bg=CARD)
    f.pack(fill="x", pady=8)
    return f



```

```python
# ---------- Student ----------
def student_page():
    clear()
    title("Student Management", "Manage students and their grades")

    f = card()
    tk.Label(f, text="Student", bg=CARD, fg=TEXT,
             font=("Segoe UI", 15, "bold")).pack(pady=10)

    button(f, "＋ Add Student", add_student)
    button(f, "＋ Add Grade", add_grade, GREEN)
    button(f, "⌕ Search Student", search_student)
    button(f, "✎ Update Student", update_student)
    button(f, "✕ Delete Student", delete_student, RED)


def add_student():
    clear()
    title("Add Student", "Create a new student")

    f = card()
    id_e = field(f, "Student ID")
    name_e = field(f, "Name")
    email_e = field(f, "Email")

    def save():
        if not all([id_e.get(), name_e.get(), email_e.get()]):
            messagebox.showerror("Error", "Please fill all fields")
            return
        system.add_student(student(id_e.get(), name_e.get(), email_e.get()))
        messagebox.showinfo("Success", "Student added successfully")
        student_page()

    button(f, "Save Student", save, GREEN)


def add_grade():
    clear()
    title("Add Grade", "Enter grades from 0 to 100")

    f = card()
    id_e = field(f, "Student ID")
    a = field(f, "Arabic")
    e = field(f, "English")
    m = field(f, "Math")
    c = field(f, "Chemistry")
    p = field(f, "Physics")

    def save():
        try:
            grades = list(map(int, [a.get(), e.get(), m.get(), c.get(), p.get()]))
            if any(x < 0 or x > 100 for x in grades):
                raise ValueError
        except ValueError:
            messagebox.showerror("Error", "Grades must be between 0 and 100")
            return

        for s in system._GradingSystem__students:
            if s.get_id() == id_e.get():
                s.add_grade(*grades)
                messagebox.showinfo("Success", "Grades added successfully")
                student_page()
                return

        messagebox.showerror("Error", "Student not found")

    button(f, "Save Grades", save, GREEN)


def search_student():
    clear()
    title("Search Student", "Find a student by ID")

    f = card()
    id_e = field(f, "Student ID")

    def search():
        for s in system._GradingSystem__students:
            if s.get_id() == id_e.get():
                output(s.get_info())
                return
        output("Student Not Found")

    button(f, "Search", search)


def update_student():
    clear()
    title("Update Student", "Change student information")

    f = card()
    id_e = field(f, "Student ID")
    name_e = field(f, "New Name")
    email_e = field(f, "New Email")

    def update():
        found = False
        if name_e.get():
            found = system.update_Name_Student_By_ID(id_e.get(), name_e.get()) != "Not Found"
        if email_e.get():
            found = system.update_Email_Student_By_ID(id_e.get(), email_e.get()) != "Not Found"

        if found:
            messagebox.showinfo("Success", "Student updated")
            student_page()
        else:
            messagebox.showerror("Error", "Student not found")

    button(f, "Update", update, GREEN)


def delete_student():
    clear()
    title("Delete Student", "Remove a student from the system")

    f = card()
    id_e = field(f, "Student ID")

    def delete():
        r = system.Delete_Student_By_ID(id_e.get())
        if r == "Student Deleted":
            messagebox.showinfo("Success", r)
            student_page()
        else:
            messagebox.showerror("Error", r)

    button(f, "Delete Student", delete, RED)


# ---------- Teacher ----------
def teacher_page():
    clear()
    title("Teacher Management", "Manage teachers")

    f = card()
    tk.Label(f, text="Teacher", bg=CARD, fg=TEXT,
             font=("Segoe UI", 15, "bold")).pack(pady=10)

    button(f, "＋ Add Teacher", add_teacher)
    button(f, "⌕ Search Teacher", search_teacher)
    button(f, "✎ Update Teacher", update_teacher)
    button(f, "✕ Delete Teacher", delete_teacher, RED)


def add_teacher():
    clear()
    title("Add Teacher", "Create a new teacher")

    f = card()
    id_e = field(f, "Teacher ID")
    name_e = field(f, "Name")
    email_e = field(f, "Email")
    subject_e = field(f, "Subject")

    def save():
        if not all([id_e.get(), name_e.get(), email_e.get(), subject_e.get()]):
            messagebox.showerror("Error", "Please fill all fields")
            return
        system.add_teacher(Teacher(id_e.get(), name_e.get(),
                                   email_e.get(), subject_e.get()))
        messagebox.showinfo("Success", "Teacher added")
        teacher_page()

    button(f, "Save Teacher", save, GREEN)


def search_teacher():
    clear()
    title("Search Teacher", "Find a teacher by ID")

    f = card()
    id_e = field(f, "Teacher ID")

    def search():
        for t in system._GradingSystem__teachers:
            if t.get_id() == id_e.get():
                output(t.get_info())
                return
        output("Teacher Not Found")

    button(f, "Search", search)


def update_teacher():
    clear()
    title("Update Teacher", "Change teacher information")

    f = card()
    id_e = field(f, "Teacher ID")
    name_e = field(f, "New Name")
    email_e = field(f, "New Email")
    subject_e = field(f, "New Subject")

    def update():
        found = False
        if name_e.get():
            found = system.update_Name_Teacher_By_ID(id_e.get(), name_e.get()) != "Not Found"
        if email_e.get():
            found = system.update_Email_Teacher_By_ID(id_e.get(), email_e.get()) != "Not Found"
        if subject_e.get():
            found = system.update_Subject_Teacher_By_ID(id_e.get(), subject_e.get()) != "Not Found"

        if found:
            messagebox.showinfo("Success", "Teacher updated")
            teacher_page()
        else:
            messagebox.showerror("Error", "Teacher not found")

    button(f, "Update", update, GREEN)


def delete_teacher():
    clear()
    title("Delete Teacher", "Remove a teacher")

    f = card()
    id_e = field(f, "Teacher ID")

    def delete():
        r = system.Delete_Teacher_By_ID(id_e.get())
        if r == "Teacher Deleted":
            messagebox.showinfo("Success", r)
            teacher_page()
        else:
            messagebox.showerror("Error", r)

    button(f, "Delete Teacher", delete, RED)


```

```python
# ---------- Management ----------
def management_page():
    clear()
    title("Management", "Overview of your grading system")

    for text, color, func in [
        ("👨‍🎓  Show All Students", BLUE, show_students),
        ("👨‍🏫  Show All Teachers", GREEN, show_teachers)
    ]:
        f = card()
        button(f, text, func, color)


def show_students():
    data = ""
    for s in system._GradingSystem__students:
        data += "━━━━━━━━━━━━━━━━━━━━\n" + s.get_info() + "\n"
    output(data or "No students found")


def show_teachers():
    data = ""
    for t in system._GradingSystem__teachers:
        data += "━━━━━━━━━━━━━━━━━━━━\n" + t.get_info() + "\n"
    output(data or "No teachers found")



```

```python
# ---------- Main Window ----------
root = tk.Tk()
root.title("Grading System")
root.geometry("1000x650")
root.configure(bg=BG)

# Sidebar
sidebar = tk.Frame(root, bg=SIDE, width=220)
sidebar.pack(side="left", fill="y")
sidebar.pack_propagate(False)

tk.Label(sidebar, text="GRADING\nSYSTEM", bg=SIDE, fg=TEXT,
         font=("Segoe UI", 20, "bold")).pack(pady=35)

for text, command in [
    ("🎓  Students", student_page),
    ("👨‍🏫  Teachers", teacher_page),
    ("⚙  Management", management_page)
]:
    tk.Button(sidebar, text=text, command=command, bg=SIDE, fg=TEXT,
              activebackground="#1E293B", activeforeground=TEXT,
              relief="flat", bd=0, anchor="w",
              font=("Segoe UI", 11, "bold"),
              cursor="hand2", padx=25, pady=13).pack(fill="x", padx=10)

tk.Label(sidebar, text="OOP Grading System\nPython + Tkinter",
         bg=SIDE, fg=MUTED, font=("Segoe UI", 9)).pack(side="bottom", pady=25)

# Main
main = tk.Frame(root, bg=BG)
main.pack(side="left", fill="both", expand=True)

content = tk.Frame(main, bg=BG)
content.pack(fill="both", expand=True, padx=35, pady=15)

result = tk.Text(main, height=8, bg="#020617", fg=TEXT,
                 insertbackground=TEXT, relief="flat",
                 font=("Consolas", 10))
result.pack(fill="x", padx=35, pady=(0, 20))
result.config(state="disabled")

student_page()
root.mainloop()
```
